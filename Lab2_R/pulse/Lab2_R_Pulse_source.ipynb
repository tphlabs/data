# Lab2 Resistance experiment

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from scipy.interpolate import interp1d
import requests
```

```python
names = range(0, 25)
fig, ax = plt.subplots(5, 5, figsize=(16, 8), sharey=True, sharex=True)

for i, name in enumerate(names):
  url = f'{name}.csv'

  try:
    t, vx,  = np.genfromtxt(url, delimiter=',', skip_header=1).T
  except FileNotFoundError:
    # Code to execute if the file is not found (skip the read operation)
    print(f"File not found, skipping: {url}")
    continue

  # Calculate row and column indices for the 2x6 subplot grid
  row_idx = i // 5 # Integer division to get row (0 or 1)
  col_idx = i % 5  # Modulo to get column (0-5)

  axx = ax[row_idx, col_idx]
  axx.plot(t*1e3, vx*1e3)
  axx.set_title(f'sample {name}', pad=-500, fontsize=10)

  #axx.set_xlabel('time, ms')
  #axx.set_ylabel('voltage, mV')
  #axx.set_ylim(-500, 2000)


  #axx.legend() # UserWarning: No artists with labels found to put in legend.
  axx.grid()
```
